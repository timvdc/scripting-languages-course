## Functions

In the lecture on Wednesday, we started exploring functions. Python, like most other programming languages, has functions that allow us to associate a name with a procedure. The procedure is just a series of commands that can be executed jointly by calling the function. Usually, a function takes some kind of input and computes an output (although some functions do not have any inputs or outputs). An easy example is a function that finds the area of a rectangle:

````python
def rect_area(width, height):
    return width * height

rect_area(4, 6)
````

Let’s go through this function definition bit by bit. The `def` is a special, fixed keyword that tells Python we are defining a function. Next comes the **name** of the function, which we get to choose. In this case, we chose to call it `rect_area`, but it could have been something else as well, like `area` or `rectangle_area`, as long as it does not contain spaces, nor starts with a digit, and as long as you use it consistently throughout your code.

Next come the **parameters** of the function, which are names that stand in for the inputs we will give our function later when we actually know what they are (the actual values that we are giving as input to the function are called the **arguments**). The colon tells Python that we’re ready to begin describing what our function needs to compute, and we begin the next line with four spaces.

The `return` tells Python that we want to supply the result of the following expression as the output of this function.

Note that the body of the function is indented. Python needs this space in order to tell where the function ends. You may use either tabs or spaces to create this space, as long as you are consistent. It is common practice to use four spaces by default.

After the function has been defined, we can call it to find its output given some inputs. As we can see for our example, the syntax for this is just the name of the function followed by its arguments, separated by commas.

The function is now defined, and you can call it anytime you need it (provided that you've run the previous code cell).

````python
rect_area(2,3.5)
````

### Exercise 3: counting *e*'s

First, let's apply some of the techniques we already know about. Write some code that counts how many times the letter *e* occurs in `text`. Make use of a `for` loop and an `if` statement.

````python
# insert your code here

number_of_es = 0

for char in text:
    if char == 'e':
        number_of_es += 1

print(number_of_es)
````

Using this method, counting the number of occurrences of the letter *e* can be written as a single line of code:

### Exercise 4: counting *in*s

Write some code that defines the variable `number_of_hits` and counts how often the word *in* (assigned to `item_to_count`) occurs in the list of words called `words`.

````python
words = text.split()
number_of_hits = 0
item_to_count = 'in'

# insert your code here
for word in words:
    if word == item_to_count:
        number_of_hits += 1
print(number_of_hits)
````

We will go through the previous exercise step by step. We would like to know how often the preposition *in* occurs in our text. As a first step we will split the string `text` into a list of words:

````python
words = text.split()
````

Next we define a variable `number_of_hits` and set it to zero.

````python
number_of_hits = 0
````

The final step is to loop over all words in `words` and add 1 to `number_of_ins` if we find a word that is equal to `in`:

````python
item_to_count = 'in'
for word in words:
    if word == item_to_count:
        number_of_hits += 1
print(number_of_hits)
````

Now, say we would like to know how often the word *of* occurs in our text. We could adapt the previous lines of code to search for the word *of*, but what if we also would like to count the number of times *the* occurs, and *house* and *had* and... It would be really cumbersome to repeat all these lines of code for each particular search term we have. As you might have guessed, we are going to create a function that is able to perform the process for any word we give it as input.  We want to write a function called `count_in_list`. It takes two arguments: (1) the object we would like to count and (2) the list in which we want to count that object. Our function looks like this:

````python
def count_in_list(item_to_count, list_to_search):
    number_of_hits = 0
    for item in list_to_search:
        if item == item_to_count:
            number_of_hits += 1
    return number_of_hits
````

Let's go through our function one more time, just to make sure you really understand all of it.

1. First we define a function using `def` and give it the name `count_in_list` (line 1);
2. This function takes two arguments: `item_to_count` and `list_to_search` (line 1);
3. Within the function, we define a variable `number_of_hits` and assign to it the value zero (since at that stage we haven't found anything yet (line 2));
4. We loop over all words in `list_to_search` (line 3);
5. If we find a word that is equal to `item_to_count` (line 4), we add 1 to `number_of_hits` (line 5);
6. Return the result of `number_of_hits` (line 6).


Let's test our little function. We will first count how often the word *an* occurs in our list of words `words`.

````python
print(count_in_list('an', words))
````

## Dictionaries

In the lecture on Wednesday, we also introduced the Python dictionary. A dictionary is another data structure which can contain multiple items. However, its items consist of keys and values, and the dictionary allows you to look up a *value* given its *key*. It is called a dictionary because internally it is organized in such a way that looking up an item can be done quickly, just as an actual dictionary. Although the name dictionary is a metaphor and the keys and values can be anything, it happens to be very useful when working with words! Let's define one:

````python
my_dict = {'book': 'physical objects consisting of a number of pages bound together',
           'sword': 'a cutting or thrusting weapon that has a long metal blade',
           'pie': 'dish baked in pastry-lined pan often with a pastry top'}
````

Take a close look at the new syntax. Notice the curly braces and the colons. Keys are located at the left side of the colon; values at the right side. Key-value pairs (items) are separated by commas. For readability each item is on a separate line, but we could also write the whole dictionary on a single line. (In general, between parentheses and brackets you may always add extra whitespace and newlines to improve readability). To look up the value of a given key, we 'index' the dictionary using that key:

````python
print(my_dict['sword'])
````

We say 'index', because we use the same syntax with square brackets when indexing lists or strings. The difference is that we don't use a position number to index a dictionary, but a key. Like lists, dictionaries are mutable which means we can add, remove, and change entries.

Instead of storing definitions as in an actual dictionary, we can also store other values, such as frequencies. Let's define an empty dictionary and add some words and frequencies to it. Remark the syntax to add a new entry:

````python
frequencies = {}
frequencies['pride'] = 8
frequencies['prejudice'] = 9
````

## A more general count function

Our function `count_in_list` is a convenient piece of code allowing us to easily count how often certain items occur in a given list. Now what if we would like to find out for all words in our text how often they occur. Then it would be still quite cumbersome to call our function for each unique word. We would like to have a function that takes as argument a particular list and counts for each unique item in that list how often it occurs. In the lecture on Wednesday, we explored a more general way of counting words in a text by making use of a dictionary. Let's create a function that implements that idea.

We will use a dictionary to write the function `counter` that takes as argument a list and returns a `dict` with a key for each unique item and a value corresponding to the number of times it occurs in the list. We will first write some code without the function declaration. If that works, we will add it, just as before, to the body of a function.

We start with defining a variable `counts` which is an empty dictionary:

````python
counts = {}
````

Next we will loop over all words in our list `words`. For each word, we check whether the dictionary already contains it. If not, we add the word to the dictionary and initialize the count to 0. Then, we add 1 to its value.

````python
for word in words:
    if word not in counts:
        counts[word] = 1
    else:
        counts[word] += 1
````

````python
counts
````

Now that our code is working, we can add it to a function. We define the function `counter` using the `def` keyword. It takes one argument (`list_to_search`).

````python
def counter(list_to_search):
    counts = {}
    for word in list_to_search:
        if word not in counts:
            counts[word] = 1
        else:
            counts[word] += 1
    return counts
````

Let's try out our new function:

````python
counter(words)
````

### Exercise 7: character frequencies

Write a function `char_freq()` that takes a string and builds a frequency list of the characters contained in it. Represent the frequency list as a Python dictionary.
   Try it with something like `char_freq("abbabcbdbabdbdbabababcbcbab")`.

````python
# insert your code here

def char_freq(input_string):
    count_chars = {}
    for char in input_string:
        if char not in count_chars:
            count_chars[char] = 0
        count_chars[char] += 1
    return count_chars

# Note that the way we add elements to our dictionary is different compared to
# the counter function we defined before; do you see how this works?

char_freq("abbabcbdbabdbdbabababcbcbab")
````

### Exercise 8: sorted character frequencies

Write a function `char_freq_table()` that takes a string as argument, builds a frequency list of the characters contained in it, and prints an alphabetically sorted and nicely formatted character frequency table to the screen.

Hint: a simple way to format a table is to use `'\t'` to insert a tab-character with variable length between the columns.

```
In [1] char_freq_table(text)
Out
a     12
b      9
c    112
...
```

````python
# insert your code here

def char_freq_table(input_string):
    outputString = ''
    char_freqs = char_freq(input_string)
    sorted_chars_list = list(char_freqs.keys())
    sorted_chars_list.sort()
    for char in sorted_chars_list:
        outputString += char + '\t' + str(char_freqs[char]) + '\n'
    return outputString

print(char_freq_table("abbabcbdbabdbdbabababcbcbab"))
````

### Exercise 9: word frequencies above a threshold

Write a function that counts the frequency of words in a given text, and returns all the words that occur with a frequency of at least $n$, where $n$ is an integer. Make sure that your can provide $n$ as an argument to your function.

````python
# insert your code here

def frequencies_threshold(text, threshold):
    words = text.split()
    counts = {}
    for word in words:
        if word not in counts:
            counts[word] = 1
        else:
            counts[word] += 1
    counts_threshold = {}
    for word, freq in counts.items():
        if freq >= threshold:
            counts_threshold[word] = freq
    return counts_threshold

frequencies_threshold(text, 5)
````

### Exercise 10: count upper and lower case

Write a function that takes a string as input, and computes the number of upper case letters and lower case letters that are present in the string.



````python
# insert your code here

def count_cases(input_string):
    upper = 0
    lower = 0

    for char in input_string:
        if char.isupper():
            upper += 1
        elif char.islower():
            lower += 1
    return upper, lower

count_cases('ThIS A teST SEntEncE')
````

### Challenge 1: words by frequency

The function `counter` gives you a dictionary that maps each word to its frequency. Write a function `words_by_frequency` that takes such a dictionary and turns it around: it returns a dictionary that maps each frequency to the *list* of words that have that frequency. For example, if `in` and `the` both occur 3 times, the key `3` should map to `['in', 'the']` (in any order).

Test it on `counter(text.split())`, and print the words that occur exactly twice.

````python
def words_by_frequency(frequencies):
    by_frequency = {}
    for word, frequency in frequencies.items():
        if frequency not in by_frequency:
            by_frequency[frequency] = [word]
        else:
            by_frequency[frequency].append(word)
    return by_frequency

frequencies = counter(text.split())
by_frequency = words_by_frequency(frequencies)
print(by_frequency[2])
````

### Challenge 2: word pairs

A *bigram* is a pair of words that occur next to each other in a text. Write a function `count_bigrams` that takes a list of words and returns a dictionary that maps each bigram to the number of times it occurs. Use a tuple `(word, next_word)` as the key. (Hint: loop over the positions `0` up to `len(words) - 1` with `range`, so that you can look at the word at position `i` and the word at position `i + 1`.)

Then print the three most frequent bigrams of the text, with their counts. Use the trick from the lecture: build a list of `(count, bigram)` tuples and sort it.

````python
def count_bigrams(words):
    bigrams = {}
    for i in range(len(words) - 1):
        bigram = (words[i], words[i + 1])
        if bigram not in bigrams:
            bigrams[bigram] = 1
        else:
            bigrams[bigram] += 1
    return bigrams

bigrams = count_bigrams(text.split())
sorted_bigrams = []
for bigram, count in bigrams.items():
    sorted_bigrams.append((count, bigram))
sorted_bigrams.sort(reverse=True)
print(sorted_bigrams[:3])
````

### Challenge 3: word length profile

Write a function `length_profile` that takes a list of words and returns a dictionary that maps each word length to the number of words that have that length. Then print the profile as a histogram, one line per length, from the shortest to the longest length, in the following format (the counts are made up):

```
 1  **
 2  ********
 3  *****************
```

Hint: a list of the keys of a dictionary can be sorted; see the lecture slides. To align the numbers, you can use an f-string with a width, e.g. `f'{length:2}'` prints the number in a field of two characters.

````python
def length_profile(words):
    profile = {}
    for word in words:
        length = len(word)
        if length not in profile:
            profile[length] = 1
        else:
            profile[length] += 1
    return profile

profile = length_profile(text.split())
lengths = list(profile.keys())
lengths.sort()
for length in lengths:
    print(f'{length:2}  ' + '*' * profile[length])
````
