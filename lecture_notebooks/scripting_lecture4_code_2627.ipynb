### Defining functions

A function bundles a few statements under a name, so you can reuse them.

```python
def print_refrain():
    print('Twinkle, twinkle, little star,')
    print('How I wonder what you are!')

def print_stanza():
    print('Up above the world so high,')
    print('Like a diamond in the sky.')
```

### Calling functions

Defining a function does not run it; calling it (with parentheses) does.

```python
print_refrain()
```

```python
print_stanza()
```

### Re-using functions

Functions can call other functions.

```python
def print_full_lyrics():
    print_refrain()
    print_stanza()
    print_refrain()

print_full_lyrics()
```

### Function arguments

Parameters make a function flexible: the same code works for any input.

```python
def display_greeting(name):
    print('Hello', name)

display_greeting('David')
```

### Returning results

`return` hands a result back to the caller, so you can store it in a variable and keep working with it.

```python
def add_smiley(sentence):
    new_sentence = sentence + ' :)'
    return new_sentence

result = add_smiley('Programming is fun')
print(result)
```

A function without a `return` statement returns the special value `None`:

```python
def say_hello():
    print('Hello')

a = say_hello()
print(a)
```

```python
def even_odd(number):
    if number % 2 == 0:
        return 'even'
    else:
        return 'odd'

result = even_odd(5)
print(result)
```

### Dictionaries

A dictionary maps keys to values, like a phone book maps names to numbers.

```python
coworkers = {'Michael': 123, 'Alice': 456, 'Kate': 789}
print(coworkers['Alice'])
coworkers['Bob'] = 124
print(coworkers)
```

### Working with dictionaries

Looking up, adding, testing for a key, counting entries, and removing an entry.

```python
coworkers = {'Michael': 123, 'Alice': 456}
print(coworkers['Alice'])
coworkers['Bob'] = 124
print('Alice' in coworkers)
print(len(coworkers))
del coworkers['Michael']
print(coworkers)
print(coworkers.keys())
print(coworkers.values())
print(coworkers.items())
```

### Using a dictionary to count

The counting idiom: if the key is new, set it to 1, otherwise add 1. You will use this constantly.

```python
word = 'interdisciplinary'
character_counter = {}
for char in word:
    if char not in character_counter:
        character_counter[char] = 1
    else:
        character_counter[char] += 1
print(character_counter)
```

### Counting words

The same idiom applied to the words of a sentence (the sentence is from Nabokov).

```python
sentence = """
As he crossed toward the pharmacy at the corner he involuntarily
turned his head because of a burst of light that had ricocheted from
his temple and saw with that quick smile with which we greet a
rainbow or a rose a blindingly white parallelogram of sky being
unloaded from the van a dresser with mirrors across which as across a
cinema screen  passed a flawlessly clear reflection of boughs sliding
and swaying not arboreally but with a human vacillation produced by
the nature of those who were carrying this sky these boughs this
gliding facade
"""
```

```python
word_list = sentence.lower().split()
word_counter = {}
for word in word_list:
    if word not in word_counter:
        word_counter[word] = 1
    else:
        word_counter[word] += 1
print(word_counter)
```

### Looping over dictionaries

Looping over a dictionary gives you its keys; `.items()` gives key and value at once.

```python
data = {'a': 0, 'b': 1}
for key in data:
    print(key)
    print(data[key])
```

```python
for key, value in data.items():
    print(key)
    print(value)
```

### Tuples

A tuple is like a list, but immutable; unlike lists, tuples can be used as dictionary keys.

```python
couple = ('John', 'Mary')
years_married = {('John', 'Mary'): 3, ('Alice', 'Bob'): 5}
print(years_married[couple])
```

Indexing works as for lists; unpacking assigns the items to separate variables; tuples are compared item by item.

```python
couple = ('John', 'Mary')
print(couple[0])
name, partner = couple
print(partner)
print((3, 'the') < (5, 'a'))
print((3, 'a') < (3, 'the'))
```
