### Practical application: testing code

A comparison with `==` gives a Boolean value (`True` or `False`); handy for checking your own computations.

```python
x = 3 * 3
print(x == 3 + 3 + 3)
```

### The if-statement

The indented block only runs when the condition is `True`.

```python
score = 11
if score >= 10:
    print('you passed the exam')
```

```python
score = 7
if score >= 10:
    print('you passed the exam')
# nothing is printed: the condition is False
```

The next cell produces an error on purpose; that is exactly the point of the example. You cannot compare a string to a number.

```python
score = '12'
if score >= 10:
    print('you passed the exam')
```

### Optional extensions of the if-statement

`else` catches everything the condition missed; `elif` adds extra conditions in between.

```python
score = 7
if score >= 10:
    print('you passed the exam')
else:
    print('you should have studied more')
```

```python
score = 18
if score > 16:
    print('you did really well!')
elif score >= 10:
    print('you passed the exam')
else:
    print('you should have studied more')
```

### Complex conditions

Conditions can be combined with `and`, `or`, and `not`.

```python
score = 12
if score >= 10 and score < 15:
    print('you passed, but there is room for improvement')
```

### What counts as True and False?

An empty list counts as `False`, a non-empty list as `True`; the three conditions below are equivalent.

```python
shoppinglist = ['bread']

if len(shoppinglist) != 0:
    print('go to the supermarket')
if len(shoppinglist):
    print('go to the supermarket')
if shoppinglist:
    print('go to the supermarket')
```

### Converting a decision tree into if-statements

Should you stay at the reception? Two versions of the same decision logic.

```python
awkward = False
free_food = True

if awkward:
    if free_food:
        print('Stay')
    else:
        print('Leave')
else:
    print('Stay')
```

```python
awkward = False
free_food = True

if not awkward:
    print('Stay')
elif free_food:
    print('Stay')
else:
    print('Leave')
```

```python
awkward = False
free_food = True

if awkward and not free_food:
    print('Leave')
else:
    print('Stay')
```

### Repeating statements: for-loops

The loop below replaces writing out the print statement once per item.

```python
shoppinglist = ['a', 'b', 'c']
for item in shoppinglist:
    print(item)
```

### Looping over sequences

You can loop over any sequence, including the characters of a string.

```python
for letter in 'word':
    print(letter)
```

```python
for number in [0, 1, 2]:
    print(number)
```

### Nesting loops inside loops

```python
for number in [1, 2]:
    for letter in 'abc':
        print(number, letter)
```

### Looping without a list

`range(4)` generates the numbers 0 up to (but not including) 4.

```python
for number in range(4):
    print(number)
```

### While-loop

A while-loop keeps repeating as long as its condition holds.

```python
countdown = 10
while countdown > 0:
    print(countdown)
    countdown -= 1
print('Liftoff!')
```

### List comprehensions

A compact way to build a new list from an existing one. First the loop version, then the equivalent comprehension.

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

fruits_with_letter_a = []
for x in fruits:
    if "a" in x:
        fruits_with_letter_a.append(x)
print(fruits_with_letter_a)
```

```python
fruits_with_letter_a = [x for x in fruits if "a" in x]
print(fruits_with_letter_a)
```

Comprehensions can filter and transform:

```python
print([i for i in range(10)])
print([i for i in range(10) if i < 5])
print([i for i in range(10) if i % 2 == 0])
print([i*2 for i in range(10) if i >= 5])
```

### Example: counting letters

How many times does the letter 'a' occur in 'banana'?

```python
word = 'banana'
count = 0
for letter in word:
    if letter == 'a':
        count += 1
print(count)
```

### Reversing text

First with a loop, then with the slicing shortcut `[::-1]`.

```python
string = 'I am testing'
result = ''
for letter in string:
    result = letter + result
print(result)
```

```python
print(string[::-1])
```

### Palindromes

A palindrome reads the same backwards.

```python
string = 'radar'
backwards = string[::-1]
print(backwards == string)
```

### Overlapping lists?

Do two lists share an element? First with two nested loops, then more simply with the `in` operator.

```python
list1 = [1, 2, 3]
list2 = [1, 4, 6]
overlapping = False

for elem1 in list1:
    for elem2 in list2:
        if elem1 == elem2:
            overlapping = True
print(overlapping)
```

```python
overlapping = False
for elem1 in list1:
    if elem1 in list2:
        overlapping = True
print(overlapping)
```
