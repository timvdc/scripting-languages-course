### Python as a calculator

Python respects the usual order of operations; parentheses change it.

```python
1 + 2 * 3
```

```python
(1 + 2) * 3
```

### Numeric data types

Division gives a `float`; integer division (`//`) gives an `int`.

```python
print(5 / 2)
print(type(5 / 2))
print(type(5 // 2))
```

### Variables

A variable stores a value, so you can use it later.

```python
a = 5 / 2
a
```

```python
a + 0.5
```

```python
a = 2
print(a)
a = 3
print(a)
```

### Updating variables

A variable can be updated based on its current value; `a += 1` is shorthand for `a = a + 1`.

```python
a = 2
a = a + 1
print(a)

a = 2
a += 1
print(a)
```

### Strings

```python
text = 'This is a string of characters.'
type(text)
```

### Indexing: extracting one element

Indexing starts at 0; negative indices count from the end.

```python
name = 'Marcel'
print(name[0])
print(name[-1])
```

### Slicing: extracting a range of elements

The start index is included, the end index is not.

```python
name = 'Marcel'
print(name[1:3])
print(name[:2])
print(name[1:])
```

### Length

```python
name = 'Marcel'
len(name)
```

### Format strings

Three ways to combine text and values in a `print`; the f-string (last one) is the most readable.

```python
x = 1
y = 2
z = x + y

print('The sum of ' + str(x) + ' and ' + str(y) + ' is ' + str(z) + '.')
print('The sum of', x, 'and', y, 'is', z, '.')
print(f'The sum of {x} and {y} is {z}.')
```

### Lists

```python
numbers = [0, 1, 2]
names = ['John', 'Mary']
```

### Changing items

Lists are mutable: you can change their elements.

```python
names = ['John', 'Mary']
names[1] = 'Alice'
names
```

Strings, on the other hand, are immutable. The next cell produces an error on purpose; that is exactly the point of the example.

```python
name = 'Maria'
name[0] = 'D'
```

### Adding items

```python
names = ['John', 'Mary']
names.append('Alice')
names
```

```python
friends = ['John', 'Mary']
enemies = ['Alice', 'Bob']
friends + enemies
```

### Removing items

```python
names = ['John', 'Mary']
names.remove('John')
names
```

```python
names = ['John', 'Mary', 'Alice', 'Bob', 'Max']
print(names.pop(3))  # removes 4th item
print(names.pop())   # removes last item
print(names)
```

Slicing also works on lists, and gives you a new list:

```python
names = ['John', 'Mary', 'Alice', 'Bob']
friends = names[:2]
friends
```

### Sorting

`sorted()` returns a new sorted list and leaves the original untouched; the `.sort()` method sorts the list in place.

```python
letters = ['b', 'd', 'a', 'c']
print(sorted(letters))
print(letters)
letters.sort()
print(letters)
```

### Nested lists: lists inside of lists

```python
ages = [['John', 34], ['Mary', 19], ['Max', 62]]

name_and_age = ages[0]
print(name_and_age)
print(name_and_age[1])
print(ages[0][1])
```

You can also unpack a two-element list into two variables at once:

```python
name, age = name_and_age
print(name)
print(age)
```
