The archive contains a directory `books` with a number of well-known English novels (plus two other directories that we use in the practical session).

### Text encoding: specifying the encoding

Note: the next cell produces an error on purpose. If you try to read a file that is not in UTF-8, Python raises a `UnicodeDecodeError`.

```python
# This will cause an error if the file
# is not in UTF-8:
with open('document_latin1.txt') as in_file:
    for line in in_file:
        print(line)
```

```python
# Fix: specify the encoding
with open('document_latin1.txt', encoding='latin1') as in_file:
    for line in in_file:
        print(line)
```

### Finding files by pattern

`glob` matches a set of files with a search pattern: `*` means "match zero or more characters", `?` means "match exactly one character".

```python
from glob import glob

glob('data/*.txt')
```

```python
glob('books/austen-*.txt')
```

The same patterns work on the command line:

```python
!ls data/*.txt
```

### Processing multiple files

```python
from glob import glob

for filename in glob('data/*.txt'):
    with open(filename) as in_file:
        text = in_file.read()
        word_count = len(text.split())
        print(f'{filename}: {word_count} words')
```

### Sorting with a key function

```python
names = ['Charlie', 'Alice', 'Bob']
sorted(names)
```

```python
# Sort by length instead of alphabetically:
sorted(names, key=len)
```

```python
sorted(names, key=len, reverse=True)
```

### Saving Python objects with pickle

```python
translation_dictionary = {'red': 'rouge', 'green': 'vert'}

import pickle
with open('translations.pickle', 'wb') as out_file:
    pickle.dump(translation_dictionary, out_file)

with open('translations.pickle', 'rb') as in_file:
    my_translations = pickle.load(in_file)

my_translations
```
