### Dog class

Our own class. The constructor `__init__` is called when an object is created; `self` refers to the current instance.

```python
class Dog:

    def __init__(self, breed, name):
        self.breed = breed
        self.name = name
        self.stomach = []

    def bark(self):
        print('Woof woof')

    def eat(self, foodstuff):
        self.stomach.append(foodstuff)
```

```python
lassie = Dog('collie', 'Lassie')
lassie.bark()
```

```python
lassie.eat('sausage')
lassie.stomach
```

```python
lassie.name
```

```python
type(lassie)
```

### The parameter `self`

Every method has `self` as its first parameter: the instance the method is called on. Python fills it in. The two calls below do the same thing:

```python
lassie.eat('bone')
lassie.stomach
```

```python
Dog.eat(lassie, 'biscuit')
lassie.stomach
```

### Many instances

Each instance has its own copy of the instance variables.

```python
lassie = Dog('collie', 'Lassie')
rex = Dog('boxer', 'Rex')
lassie.eat('sausage')
lassie.stomach
```

```python
rex.stomach
```

### Instances are values

An instance can be stored in a variable, a list or a dictionary, and passed to a function, like any other value.

```python
dogs = [lassie, rex]
for dog in dogs:
    print(dog.name, len(dog.stomach))
```

### Inheritance

A subclass inherits the attributes and methods of its parent class, and can add its own.

```python
class K9(Dog):

    def search_explosives(self):
        print('Searching..')

max_pd = K9('shepherd', 'Max')
max_pd.bark()
max_pd.eat('sausage')
max_pd.search_explosives()
```

```python
max_pd.stomach
```

### Overriding a method

A subclass can define a method with the same name as the parent class; the new definition is used. `super()` gives access to the parent's version.

```python
class K9(Dog):

    def bark(self):
        super().bark()
        print('(in a police voice)')

    def search_explosives(self):
        print('Searching..')

max_pd = K9('shepherd', 'Max')
max_pd.bark()
```

### An elaborate example: artworks

```python
class Artwork:
    def __init__(self, title, artist, year_created, style):
        self.title = title
        self.artist = artist
        self.year_created = year_created
        self.style = style

    def generate_caption(self):
        return (
            f"{self.title}, created by {self.artist} "
            f"in {self.year_created}, {self.style} style"
            )
```

```python
class Painting(Artwork):
    def __init__(self, title, artist, year_created, style, canvas_type):
        super().__init__(title, artist, year_created, style)
        self.canvas_type = canvas_type

    def generate_caption(self):
        return (
            f"Painting: {super().generate_caption()}, "
            f"on {self.canvas_type}"
            )

class Sculpture(Artwork):
    def __init__(self, title, artist, year_created, style, material):
        super().__init__(title, artist, year_created, style)
        self.material = material

    def generate_caption(self):
        return (
            f"Sculpture: {super().generate_caption()}, "
            f"made of {self.material}"
            )
```

```python
mona_lisa = Painting("Mona Lisa", "Leonardo da Vinci", 1503,
                     "Renaissance", "wooden panel")

david = Sculpture("David", "Michelangelo", 1504,
                  "Renaissance", "marble")

print(mona_lisa.generate_caption())
print(david.generate_caption())
```

### Composition: a Museum has Artworks

Inheritance expresses *is a* (a Painting is an Artwork); composition expresses *has a* (a Museum has Artworks).

```python
class Museum:
    def __init__(self, name):
        self.name = name
        self.collection = []   # list of Artwork (or subclasses)

    def add(self, artwork):
        self.collection.append(artwork)

    def captions(self):
        return [a.generate_caption() for a in self.collection]
```

```python
louvre = Museum('Louvre')
louvre.add(mona_lisa)
louvre.add(david)
louvre.captions()
```
