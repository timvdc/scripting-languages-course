## Reading files

Let's now go through our Wikipedia file line by line, and put all of its words in one big list. To do so, we'll make use of the **with**-statement and the **for**-loop.

We use the `with` statement to define a file stream, accessible through the variable `in_file`. We provide the `open` function with the filename on our system (`wiki/wiki_sample_en_small.txt`) and some extra arguments to the open function (`r` indicates that we want to read from the file, and `t` indicates that we want to read the file in text mode; note that these are the default arguments, so we can omit them). By using the `with` statement, Python will automatically close the file for us once we're done reading from it (i.e., once the indented code block below the with statement is finished).

A `for` loop keeps executing a particular code block until a stop condition is met. In this case, we'll use it to loop through lines of the file one by one until we reach the end of the file. We write code indented, below the `for` loop, to tell Python what we want it to do with each element of the list. To each line, we apply the `strip` function, which removes trailing newline characters. Next we split the line into a list of words, by splitting on the whitespace (` `). Finally, we append the word list for each line to the `corpus` list; when we are done processing our file, the list will have accumulated all the words for our entire file. Note that we initially define `corpus` as an empty list, which we'll iteratively fill up using the `for` loop.

```python
corpus = []
with open('wiki/wiki_sample_en_small.txt', 'rt') as in_file:
    for line in in_file:
        line = line.strip()
        words = line.split(' ')
        corpus += words
```

We can now inspect the `corpus` list we have constructed; we can also check how many words are present in our list using the `len` function.

```python
print(corpus[:50])
print('corpus length: ', len(corpus))
```

A for loop is useful if we want to go through the input data line by line, and process it on the fly. It is also possible to read in the entire file directly as a string. For this, you can use the function `read`.

```python
with open('wiki/wiki_sample_en_small.txt', 'rt') as in_file:
    corpus_string = in_file.read()
```

## Writing results to a file

By now, you have learnt how to read files using Python from your computer, how to manipulate them, clean them up and compute frequencies of the words in a text file. Now we'll see how to write your results to a file. We have already seen how to read a text from our disk. Writing to our disk is only slightly different. The following lines of code write a single sentence to the file `first-output.txt`.

```python
with open('first-output.txt', 'w') as out_file:
    out_file.write('My first output.\n')
```

You can confirm that the file `first-output.txt` is indeed located in the folder where this practical session resides (on Google colab, you can use the folder icon on the left, or run `!ls`). As you can see it contains the line `My first output.`.

The difference with opening a file for reading is the *mode* with which we open the connection, specified by the optional second argument to `open`. Here the mode says `w`, meaning "open the file for writing". To open a file for reading, we set the mode to `r`. However, since this is Python's default setting, this is typically omitted. All the code that uses `out_file` must be indented under the `with` statement. After the `with` statement, the file is closed and the contents can be read by other programs. We write to a file with the `write` method; this is similar to the `print` function, but there are two important differences:

- `write` only accepts a single string argument, so you must format everything you write into a string.
- `print` adds an end-of-line, but we have to explicitly tell `write` that we want to end the line with `\n`.

Warning: opening a file for writing will remove any existing contents, if the file already exists.

```python
!cat first-output.txt
```

## Working with collections of files

Often we don't work with a single text file stored at our computer, but with multiple text files or entire corpora. We would like to have a way to load a corpus into Python.

Remember how to read files? Each time we had to open a file, read the contents and ensure that the file is closed. Since this is a series of steps we will often need to do, we can write a single function that does all that for us. We write a small utility function `read_file(filename)` that reads the specified file and simply returns all contents as a single string.

```python
def read_file(filename):
    """Read the contents of `filename` and return as a string."""
    with open(filename) as in_file:
        contents = in_file.read()
    return contents
```

Now, instead of having to remember the steps to read a file, we can just call the function `read_file`:

```python
text = read_file('books/blake-songs.txt')
print(text[:500])
```

In the directory `books` we have a corpus consisting of multiple files with the extension `.txt`. This corpus is a collection of English novels downloaded from the [Gutenberg](http://www.gutenberg.org) project. We want to iterate over all the filenames. You can do this using the `glob` function from the `glob` module in the Python standard library. We can import this function as follows:

```python
from glob import glob
```

After that, the `glob` function is available to use. This function takes as argument a pattern for filenames and returns all the matching files and directories:

```python
glob('books/*')
```

Notice that `glob` returns a list and we can iterate over that list. A `*` in a pattern matches zero or more characters; the result above lists all the files and directories in the `books` directory. We can make the pattern more specific and select only files ending with `.txt`. The following code will read all text files in the directory `books` and outputs the length (in characters) of each:

```python
for filepath in glob('books/*.txt'):
    text = read_file(filepath)
    print(f'{filepath} has {len(text)} characters.')
```

OK, so now we can convert the filepaths to integers corresponding to the nights of storytelling. But how will we use that to sort the corpus in chronological order? We already saw how to sort things in Python. In combination with our `get_night` function, we can use `sorted` to obtain a nicely chronologically ordered list of stories.

We start with the variable `filenames`, which already has the list of file names we want to sort. Next we call the function `sorted()` on this list and supply as keyword our function `get_night`:

```python
filenames = sorted(filenames, key=get_night)
filenames[:20]
```

A keyword argument is an argument that you specify by name. Keyword arguments must come after normal arguments.

As you can see, we now have a perfectly chronologically ordered list of filenames. As you might have guessed, the keyword argument `key=get_night` of `sorted` allows us to do this. Without this argument, Python would just sort the filenames alphabetically:

```python
filenames = glob('arabian_nights/*.txt')
filenames = sorted(filenames)
print(filenames[:20])
```

However, if we supply a function to `key`, Python will apply that function to each item before comparing it to other items during the sorting. In our case this means Python converts all filepaths to integers.
